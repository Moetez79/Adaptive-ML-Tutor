Confusion keywords

```python
confusion_keywords = [
    "i don't understand",
    "confused",
    "not clear",
    "explain again",
    "simplify",
    "what do you mean",
    "i still don't get it",
    "can you explain simply",
    "hard to understand"
]
```

Confusion detection function

```python
def detect_confusion(question):

    q = question.lower()

    confusion_keywords = [
        "don't understand",
        "confused",
        "not clear",
        "explain again",
        "simplify",
        "what do you mean",
        "still don't get it",
        "hard to understand"
    ]

    for keyword in confusion_keywords:

        if keyword in q:
            return True

    return False
```

Test Confusion Detection

```python
detect_confusion(
    "I still don't understand gradient descent"
)
```

Educational knowledge responses

```python
teaching_styles = {

    "overfitting": {

        "beginner":
        "Overfitting happens when a model memorizes training examples instead of learning general patterns.",

        "analogy":
        "Imagine a student memorizing exam answers without understanding the subject. The student performs well on known questions but fails new ones. That is similar to overfitting.",

        "concise":
        "Overfitting reduces model generalization on unseen data.",

        "detailed":
        "Overfitting occurs when a model learns noise and details from training data, causing poor performance on new unseen examples."
    },

    "gradient descent": {

        "beginner":
        "Gradient descent helps machine learning models improve step by step by reducing errors.",

        "analogy":
        "Imagine walking downhill in fog. You carefully move in the direction that goes downward most steeply. That is similar to gradient descent minimizing error.",

        "concise":
        "Gradient descent iteratively minimizes a loss function.",

        "detailed":
        "Gradient descent is an optimization algorithm that updates model parameters iteratively to minimize prediction error."
    }
}
```

Topic extraction

```python
def extract_topic(conversation_history):

    topics = [
        "overfitting",
        "gradient descent"
    ]

    text = " ".join(conversation_history).lower()

    for topic in topics:

        if topic in text:
            return topic

    return None
```

Adaptive Tutor Logic

```python
def adaptive_tutor_response(
    conversation_history,
    current_question
):

    topic = extract_topic(conversation_history)

    if topic is None:
        return "I could not identify the topic."

    confused = detect_confusion(current_question)

    if confused:

        mode = "analogy"

    elif "briefly" in current_question.lower():

        mode = "concise"

    elif "detail" in current_question.lower():

        mode = "detailed"

    else:

        mode = "beginner"

    response = teaching_styles[topic][mode]

    return {
        "topic": topic,
        "confused": confused,
        "mode": mode,
        "response": response
    }
```

Test normal interaction

```python
history = [
    "What is overfitting?"
]

question = "Can you explain it?"

adaptive_tutor_response(history, question)
```

Test confusion adaptation

```python
history = [
    "What is overfitting?"
]

question = "I still don't understand."

adaptive_tutor_response(history, question)
```

Test concise mode

```python
history = [
    "What is gradient descent?"
]

question = "Explain briefly"

adaptive_tutor_response(history, question)
```

Full conversation simulation

```python
conversation = [

    ("What is overfitting?"),

    ("Why is it bad?"),

    ("I still don't understand."),

    ("Can you explain briefly?")
]
```

```python
history = []

for q in conversation:

    history.append(q)

    result = adaptive_tutor_response(
        history,
        q
    )

    print("="*80)

    print("Student:", q)

    print("Mode:", result["mode"])

    print("Tutor:", result["response"])
```
