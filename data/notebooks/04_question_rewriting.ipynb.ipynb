```python
from transformers import T5Tokenizer, T5ForConditionalGeneration
```

Load Lightweight Rewrite Model

```python
model_name = "google/flan-t5-small"

tokenizer = T5Tokenizer.from_pretrained(model_name)

model = T5ForConditionalGeneration.from_pretrained(model_name)
```

Conversation example

```python
conversation_history = [
    "What is overfitting?",
    "Overfitting happens when a model memorizes training data.",
    
    "Why is it bad?",
    "Because it performs poorly on unseen data."
]

current_question = "How can we prevent it?"
```

Build rewrite prompt

```python
prompt = f"""
Rewrite the following conversational question into a clear standalone question.

Conversation:
{' '.join(conversation_history)}

Question:
{current_question}

Rewritten Question:
"""
```

Generate Rewritten Question

```python
inputs = tokenizer(
    prompt,
    return_tensors="pt",
    truncation=True
)

outputs = model.generate(
    **inputs,
    max_new_tokens=50
)

rewritten_question = tokenizer.decode(
    outputs[0],
    skip_special_tokens=True
)

print("Original Question:")
print(current_question)

print("\nRewritten Question:")
print(rewritten_question)
```

Create Rewriting Function

```python
def extract_topic(conversation_history):

    important_topics = [
        "overfitting",
        "gradient descent",
        "linear regression",
        "decision trees",
        "neural networks",
        "regularization",
        "classification"
    ]

    history_text = " ".join(conversation_history).lower()

    for topic in important_topics:

        if topic in history_text:
            return topic

    return "machine learning"
```

```python
def rewrite_question(conversation_history, question):

    topic = extract_topic(conversation_history)

    q = question.lower()

    if "why" in q:

        return f"Why is {topic} important in machine learning?"

    elif "how" in q:

        return f"How does {topic} work in machine learning?"

    elif "explain" in q:

        return f"Can you explain {topic} in simple terms?"

    elif "prevent" in q:

        return f"How can we prevent problems related to {topic}?"

    else:

        return f"{question} about {topic}"
```

Test Example

```python
history = [
    "What is gradient descent?",
    "Gradient descent minimizes error step by step."
]

question = "Why is it important?"

rewrite_question(history, question)
```
